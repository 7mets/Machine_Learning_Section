```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


# =========================================================
# 1. Binary Classification - 2 Classes
# =========================================================

print("=== 1. Binary Classification (2 Class) ===")

# True labels = the actual values
y_true_2 = [0, 1, 0, 1, 0, 0, 1, 1, 0, 1]

# Predicted labels = values predicted by the model
y_pred_2 = [0, 1, 0, 0, 0, 1, 1, 1, 0, 1]


# Calculate the four final evaluation metrics
accuracy_2 = accuracy_score(y_true_2, y_pred_2)
precision_2 = precision_score(y_true_2, y_pred_2)
recall_2 = recall_score(y_true_2, y_pred_2)
f1_2 = f1_score(y_true_2, y_pred_2)


# Print only the final results
print("Accuracy :", accuracy_2)
print("Precision:", precision_2)
print("Recall   :", recall_2)
print("F1-Score :", f1_2)


# Plot Confusion Matrix
cm_2 = confusion_matrix(y_true_2, y_pred_2)

plt.figure(figsize=(5, 4))

sns.heatmap(
    cm_2,
    annot=True,
    fmt='d',
    cmap='Blues',
    cbar=False
)

plt.title('Confusion Matrix - 2 Class')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

plt.show()


print("\n" + "=" * 40 + "\n")


# =========================================================
# 2. Multi-class Classification - 3 Classes
# =========================================================

print("=== 2. Multi-class Classification (3 Class) ===")

# True labels
y_true_3 = [ 0, 1, 2, 2, 1, 0,1, 2, 0, 1, 2, 0]

# Predicted labels
y_pred_3 = [ 0, 0, 2, 1, 1, 0,2, 2, 0, 1, 1, 0]


# Calculate the four final evaluation metrics
accuracy_3 = accuracy_score(y_true_3, y_pred_3)
precision_3 = precision_score(y_true_3, y_pred_3, average='weighted')
recall_3 = recall_score(y_true_3, y_pred_3, average='weighted')
f1_3 = f1_score(y_true_3, y_pred_3, average='weighted')


# Print only the final results
print("Accuracy :", accuracy_3)
print("Precision:", precision_3)
print("Recall   :", recall_3)
print("F1-Score :", f1_3)


# Plot Confusion Matrix
cm_3 = confusion_matrix(y_true_3, y_pred_3)

plt.figure(figsize=(5, 4))

sns.heatmap(
    cm_3,
    annot=True,
    fmt='d',
    cmap='Greens',
    cbar=False
)

plt.title('Confusion Matrix - 3 Class')
plt.xlabel('Predicted Label')
plt.ylabel('True Label')

plt.show()
```
