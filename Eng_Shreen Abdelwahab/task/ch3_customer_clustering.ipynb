```python
#الاسم /محمود ثروت ابراهيم صالح يحيي
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

customer_data = pd.DataFrame(
    {
        "Annual_Spending": [15000, 18000, 12000, 80000, 85000, 90000, 45000],
        "Purchase_Frequency": [2, 3, 1, 25, 28, 30, 10],
    }
)

print("=== عينة من بيانات العملاء الأصلية ===")
print(customer_data.head())
print("-" * 40)

scaler = StandardScaler()
scaled_features = scaler.fit_transform(customer_data)

best_silhouette_score = -1
best_k_value = 2

print("=== نتائج اختبار قيم K المختلفة ===")
for k_val in range(2, 5):

  temp_kmeans = KMeans(
      n_clusters=k_val, init="k-means++", n_init=10, random_state=42
  )
  cluster_labels = temp_kmeans.fit_predict(scaled_features)

  sil_avg = silhouette_score(scaled_features, cluster_labels)
  print(f"قيمة K = {k_val} | معامل السيلويت (Silhouette Score) = {sil_avg:.4f}")

  if sil_avg > best_silhouette_score:
    best_silhouette_score = sil_avg
    best_k_value = k_val

print("-" * 40)
print(f">> بناءً على التحليل، أفضل قيمة لـ K تم اختيارها هي: {best_k_value}")

final_kmeans = KMeans(
    n_clusters=best_k_value, init="k-means++", n_init=10, random_state=42
)
final_labels = final_kmeans.fit_predict(scaled_features)

customer_data["Cluster_Group"] = final_labels

print("\n=== تصنيف العملاء النهائي بعد التجميع ===")
print(customer_data)

print("\n=== المراكز النهائية للمجموعات (Centroids) ===")
print(scaler.inverse_transform(final_kmeans.cluster_centers_))
```
