```python
#الاسم/محمود ثروت ابراهيم صالح يحيي 
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

study_hours = np.array([[2], [4], [6], [8], [10]])
exam_scores = np.array([40, 50, 65, 80, 90])
regression_model = LinearRegression()
regression_model.fit(study_hours, exam_scores)

b0_intercept = regression_model.intercept_
b1_slope = regression_model.coef_[0]

print("=== نتائج تحليل الانحدار الخطي ===")
print(f"قيمة التقاطع (Intercept / Beta 0): {b0_intercept:.2f}")
print(f"قيمة الميل (Slope / Beta 1): {b1_slope:.2f}")
print(f"المعادلة النهائية للخط: y = {b0_intercept:.2f} + {b1_slope:.2f} * x")

target_hours = np.array([[7]])
estimated_score = regression_model.predict(target_hours)
print("\n=== اختبار التنبؤ ===")
print(f"الدرجة المتوقعة لطالب يذاكر {target_hours[0][0]} ساعات هي: {estimated_score[0]:.2f} درجة")

predictions = regression_model.predict(study_hours)

mae_value = mean_absolute_error(exam_scores, predictions)
mse_value = mean_squared_error(exam_scores, predictions)
rmse_value = np.sqrt(mse_value)
r2_value = r2_score(exam_scores, predictions)

print("\n=== مقاييس تقييم أداء النموذج ===")
print(f"متوسط الخطأ المطلق (MAE): {mae_value:.2f}")
print(f"متوسط مربعات الخطأ (MSE): {mse_value:.2f}")
print(f"جذر متوسط مربعات الخطأ (RMSE): {rmse_value:.2f}")
print(f"نسبة دقة وتفسير النموذج (R² Score): {r2_value * 100:.2f}%")
```
